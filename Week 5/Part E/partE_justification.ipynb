## 1 · Which variable got which channel, and why (4)

The bubble chart has four variables and two positions, so the channel budget was spent in the order the reader must compare them. **Healthcare access** (the outcome, needing the finest comparison) got the **y position**; **GDP per capita** got the **x position**, on a log scale because it spans $283–$120,219 and the spread is multiplicative. **Population** got **area**, which readers judge less precisely (position > length > angle > area) — acceptable because it is context, not the headline comparison; I passed *area* to `ax.scatter(s=...)` (diameter ratio 8.8× across an 18–1,400 pt² range, not the 589× the raw range implies) and added a size legend with real populations (1, 20, 100, 350 M). **Region** got **hue**, because it is identity, not amount, with six Okabe–Ito hues and direct labels for the four countries I discuss. Lightness was kept free for the map. In the choropleth, position was already spent on geography, so **lightness carries the rate** (one sequential hue, dark = more) and **length carries the count** in the companion bar chart.

## 2 · The parallel-coordinates decisions (4)

**Scaling: percentile rank.** `gdp_pc` has one country at 9.2× the next; under min–max 58 of 60 countries fall below 0.10 on that axis (median 0.017), under z-score the outlier sits at +7.5 SD. Rank is immune to the outlier; its cost, losing the size of gaps, is paid by printing each axis's real min and max. **Axis order: computed, not eyeballed**, by the greedy |r| rule before plotting: `healthcare_access → cases_per_100k → urban_pct → median_age → gdp_pc`. **Clutter:** all 60 lines grey, one region (Borealis, the highest mean case rate, chosen by that rule in advance) brushed. **One relationship a scatter matrix would not have shown:** 18 of the 30 above-median-access countries are *simultaneously* urban, older and low-case, and 16 sit in the opposite corner on all four — one country *profile* rather than four separate pairs. (The access–case X, r = −0.80, is the strongest single pair; Case 4 in Part A shows the file's own axis order buries it three axes apart.)

## 3 · The spatial decisions (4)

**Denominator: cases per 100,000 residents**, stated in the legend title in words, because a count map is a population map (r(cases, population) = +0.986; r(rate, population) = 0.000). **Classification: Jenks natural breaks, k = 5**, with the break edges printed in the legend; I scored equal, quantile and natural by goodness of variance fit before choosing, and the rates have real gaps, unlike Case 5's flat 7.2% spread where *no* scheme could justify five classes. **Projection: Lambert cylindrical equal-area** (standard parallel at the federation's mean latitude, 30.5°N), because colouring a polygon makes its area the encoding channel: Mercator would have inflated the far north 1.49× against 1.24× in the far south, drawing the north's ink ~20% heavier before any data was plotted. The join was checked (24/24, 0 failures); boundaries are the course's synthetic `districts.geojson` (fictional federation), read 27 Oct 2026. I used `geolib` rather than geopandas/mapclassify. Nothing was suppressed: the smallest district has 186 cases.

Published repeatedly, the count map would give the largest districts a reputation the data never supported, while directing resources away from the far north, and the cost falls on the residents of the places coloured dark, not on the mapmaker.
